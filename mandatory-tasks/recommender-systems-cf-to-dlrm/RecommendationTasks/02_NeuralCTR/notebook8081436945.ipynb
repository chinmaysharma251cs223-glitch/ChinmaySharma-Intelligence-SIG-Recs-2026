```python
import itertools
import tensorflow as tf
import matplotlib.pyplot as plt

from sklearn.model_selection import StratifiedKFold, ParameterGrid
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OrdinalEncoder
from sklearn.metrics import (
    roc_auc_score,
    average_precision_score,
    accuracy_score,
    log_loss,
    precision_score,
    recall_score,
    f1_score
)
from sklearn.base import clone


# ============================================================
# 1. REPRODUCIBILITY
# ============================================================

SEED = 42

np.random.seed(SEED)
tf.random.set_seed(SEED)


# ============================================================
# 2. LOAD DATA
# ============================================================

TRAIN_PATH = r"/kaggle/input/datasets/chinmaysharma147/dataset-task2/train.csv"
TEST_PATH = r"/kaggle/input/datasets/chinmaysharma147/dataset-task2/test.csv"

train = pd.read_csv(TRAIN_PATH)
test = pd.read_csv(TEST_PATH)

print("Train shape:", train.shape)
print("Test shape :", test.shape)


# ============================================================
# 3. SEPARATE FEATURES AND TARGET
# ============================================================

TARGET = "label"

X = train.drop(columns=[TARGET])
y = train[TARGET].astype("int32")

if TARGET in test.columns:
    X_test = test.drop(columns=[TARGET])
    y_test = test[TARGET].astype("int32")
    HAS_TEST_LABELS = True
else:
    X_test = test.copy()
    y_test = None
    HAS_TEST_LABELS = False


# ============================================================
# 4. FEATURE LISTS
# ============================================================

# Replace these with the actual column names from your dataset.

numerical_cols = [
    'integer_feature_1','integer_feature_2','integer_feature_3','integer_feature_4','integer_feature_5','integer_feature_6','integer_feature_7','integer_feature_8','integer_feature_9','integer_feature_10','integer_feature_11','integer_feature_12','integer_feature_13']

categorical_cols = [
    'categorical_feature_1','categorical_feature_2','categorical_feature_3','categorical_feature_4','categorical_feature_5','categorical_feature_6','categorical_feature_7','categorical_feature_8','categorical_feature_9','categorical_feature_10','categorical_feature_11','categorical_feature_12','categorical_feature_13','categorical_feature_14','categorical_feature_15','categorical_feature_16','categorical_feature_17','categorical_feature_18','categorical_feature_19','categorical_feature_20','categorical_feature_21','categorical_feature_22','categorical_feature_23','categorical_feature_24','categorical_feature_25','categorical_feature_26']

assert len(numerical_cols) == 13
assert len(categorical_cols) == 26


# ============================================================
# 5. PREPROCESSOR
# ============================================================

def fit_preprocessor(X_train):
    """
    Fit numerical and categorical preprocessing
    using ONLY the supplied training fold.
    """

    # Numerical
    num_imputer = SimpleImputer(strategy="median")
    X_train_num = num_imputer.fit_transform(
        X_train[numerical_cols]
    )

    scaler = StandardScaler()
    X_train_num = scaler.fit_transform(X_train_num)

    # Categorical
    cat_imputer = SimpleImputer(
        strategy="constant",
        fill_value="MISSING"
    )

    X_train_cat = cat_imputer.fit_transform(
        X_train[categorical_cols].astype(str)
    )

    encoder = OrdinalEncoder(
        handle_unknown="use_encoded_value",
        unknown_value=-1
    )

    X_train_cat = encoder.fit_transform(X_train_cat)

    # Shift so 0 can represent unknown
    X_train_cat = (
        X_train_cat.astype("int32") + 1
    )

    return (
        num_imputer,
        scaler,
        cat_imputer,
        encoder,
        X_train_num.astype("float32"),
        X_train_cat.astype("int32")
    )


def transform_data(
    X_data,
    num_imputer,
    scaler,
    cat_imputer,
    encoder
):
    """
    Transform data using preprocessing already
    fitted on the training fold.
    """

    # Numerical
    X_num = num_imputer.transform(
        X_data[numerical_cols]
    )

    X_num = scaler.transform(X_num)

    # Categorical
    X_cat = cat_imputer.transform(
        X_data[categorical_cols].astype(str)
    )

    X_cat = encoder.transform(X_cat)

    X_cat = (
        X_cat.astype("int32") + 1
    )

    return (
        X_num.astype("float32"),
        X_cat.astype("int32")
    )


# ============================================================
# 6. BUILD TENSORFLOW MODEL
# ============================================================

def build_model(
    num_features,
    categorical_cardinalities,
    hidden_units=(128, 64),
    dropout=0.2,
    learning_rate=1e-3,
    optimizer_name="adam"
):

    # Numerical input
    numerical_input = tf.keras.Input(
        shape=(num_features,),
        dtype=tf.float32,
        name="numerical"
    )

    # Categorical input
    categorical_input = tf.keras.Input(
        shape=(len(categorical_cardinalities),),
        dtype=tf.int32,
        name="categorical"
    )

    embedding_outputs = []

    for i, vocab_size in enumerate(
        categorical_cardinalities
    ):

        embedding_dim = min(
            32,
            max(4, int(np.ceil(np.sqrt(vocab_size))))
        )

        embedding = tf.keras.layers.Embedding(
            input_dim=vocab_size + 1,
            output_dim=embedding_dim,
            name=f"embedding_{i}"
        )

        x = embedding(
            categorical_input[:, i]
        )

        x = tf.keras.layers.Flatten()(x)

        embedding_outputs.append(x)

    # Combine numerical + categorical representations
    x = tf.keras.layers.Concatenate()(
        [numerical_input] + embedding_outputs
    )

    # Dense layers
    for i, units in enumerate(hidden_units):

        x = tf.keras.layers.Dense(
            units,
            activation="relu",
            name=f"dense_{i + 1}"
        )(x)

        x = tf.keras.layers.BatchNormalization(
            name=f"batch_norm_{i + 1}"
        )(x)

        if dropout > 0:
            x = tf.keras.layers.Dropout(
                dropout,
                name=f"dropout_{i + 1}"
            )(x)

    # Output
    output = tf.keras.layers.Dense(
        1,
        activation="sigmoid",
        name="click_probability"
    )(x)

    model = tf.keras.Model(
        inputs={
            "numerical": numerical_input,
            "categorical": categorical_input
        },
        outputs=output
    )

    # Optimizer
    if optimizer_name == "adam":

        optimizer = tf.keras.optimizers.Adam(
            learning_rate=learning_rate
        )

    elif optimizer_name == "batch_gd":

        optimizer = tf.keras.optimizers.SGD(
            learning_rate=learning_rate
        )

    else:
        raise ValueError(
            "optimizer_name must be 'adam' or 'batch_gd'"
        )

    model.compile(
        optimizer=optimizer,
        loss="binary_crossentropy",
        metrics=[
            tf.keras.metrics.AUC(
                name="roc_auc"
            ),
            tf.keras.metrics.AUC(
                name="pr_auc",
                curve="PR"
            ),
            tf.keras.metrics.BinaryAccuracy(
                name="accuracy"
            )
        ]
    )

    return model


# ============================================================
# 7. HYPERPARAMETER GRID
# ============================================================

param_grid = {

    "hidden_units": [
        (64,),
        (128, 64),
        (256, 128, 64)
    ],

    "dropout": [
        0.0,
        0.3
    ],

    "learning_rate": [
        1e-3,
        1e-4
    ],

    "optimizer_name": [
        "adam"
    ],

    "batch_size": [
        256
    ]
}


# ============================================================
# 8. STRATIFIED K-FOLD
# ============================================================

N_SPLITS = 5

skf = StratifiedKFold(
    n_splits=N_SPLITS,
    shuffle=True,
    random_state=SEED
)


# ============================================================
# 9. GRID SEARCH
# ============================================================

grid_results = []

param_combinations = list(
    ParameterGrid(param_grid)
)

print(
    "\nTotal hyperparameter combinations:",
    len(param_combinations)
)


for combination_number, params in enumerate(
    param_combinations,
    start=1
):

    print(
        f"\n{'=' * 70}\n"
        f"Combination {combination_number}/"
        f"{len(param_combinations)}\n"
        f"{params}\n"
        f"{'=' * 70}"
    )

    fold_scores = []

    for fold, (train_idx, val_idx) in enumerate(
        skf.split(X, y),
        start=1
    ):

        print(f"\nFold {fold}/{N_SPLITS}")

        X_fold_train = X.iloc[train_idx]
        X_fold_val = X.iloc[val_idx]

        y_fold_train = y.iloc[train_idx]
        y_fold_val = y.iloc[val_idx]

        # --------------------------------------------
        # Fit preprocessing ONLY on fold training data
        # --------------------------------------------

        (
            num_imputer,
            scaler,
            cat_imputer,
            encoder,
            X_fold_train_num,
            X_fold_train_cat
        ) = fit_preprocessor(
            X_fold_train
        )

        X_fold_val_num, X_fold_val_cat = transform_data(
            X_fold_val,
            num_imputer,
            scaler,
            cat_imputer,
            encoder
        )

        # --------------------------------------------
        # Category cardinalities
        # --------------------------------------------

        categorical_cardinalities = []

        for i in range(
            len(categorical_cols)
        ):

            cardinality = (
                X_fold_train_cat[:, i].max() + 1
            )

            categorical_cardinalities.append(
                int(cardinality)
            )

        # --------------------------------------------
        # Build model
        # --------------------------------------------

        model = build_model(
            num_features=len(numerical_cols),
            categorical_cardinalities=
                categorical_cardinalities,
            hidden_units=params["hidden_units"],
            dropout=params["dropout"],
            learning_rate=params["learning_rate"],
            optimizer_name=params["optimizer_name"]
        )

        # --------------------------------------------
        # Batch size
        # --------------------------------------------

        if params["optimizer_name"] == "batch_gd":

            batch_size = len(
                X_fold_train_num
            )

        else:

            batch_size = params["batch_size"]

        # --------------------------------------------
        # Early stopping
        # --------------------------------------------

        early_stopping = tf.keras.callbacks.EarlyStopping(
            monitor="val_loss",
            patience=5,
            restore_best_weights=True
        )

        # --------------------------------------------
        # Train
        # --------------------------------------------

        model.fit(
            {
                "numerical": X_fold_train_num,
                "categorical": X_fold_train_cat
            },
            y_fold_train,
            validation_data=(
                {
                    "numerical": X_fold_val_num,
                    "categorical": X_fold_val_cat
                },
                y_fold_val
            ),
            epochs=50,
            batch_size=batch_size,
            callbacks=[early_stopping],
            verbose=0
        )

        # --------------------------------------------
        # Validation prediction
        # --------------------------------------------

        val_prob = model.predict(
            {
                "numerical": X_fold_val_num,
                "categorical": X_fold_val_cat
            },
            verbose=0
        ).ravel()

        fold_auc = roc_auc_score(
            y_fold_val,
            val_prob
        )

        fold_scores.append(fold_auc)

        print(
            f"Fold {fold} ROC-AUC: "
            f"{fold_auc:.5f}"
        )

        del model

        tf.keras.backend.clear_session()

    # --------------------------------------------
    # Mean CV score
    # --------------------------------------------

    mean_auc = np.mean(fold_scores)
    std_auc = np.std(fold_scores)

    result = params.copy()

    result["mean_roc_auc"] = mean_auc
    result["std_roc_auc"] = std_auc

    grid_results.append(result)

    print(
        f"\nMean ROC-AUC: {mean_auc:.5f}"
        f" ± {std_auc:.5f}"
    )


# ============================================================
# 10. GRID SEARCH RESULTS
# ============================================================

results_df = pd.DataFrame(
    grid_results
).sort_values(
    "mean_roc_auc",
    ascending=False
)

print("\nGRID SEARCH RESULTS")
print(results_df.to_string(index=False))


# ============================================================
# 11. BEST PARAMETERS
# ============================================================

best_params = results_df.iloc[0].to_dict()

print("\nBEST PARAMETERS")

print(
    "Hidden units:",
    best_params["hidden_units"]
)

print(
    "Dropout:",
    best_params["dropout"]
)

print(
    "Learning rate:",
    best_params["learning_rate"]
)

print(
    "Optimizer:",
    best_params["optimizer_name"]
)


# ============================================================
# 12. FIT PREPROCESSING ON ALL PROVIDED TRAIN DATA
# ============================================================

(
    num_imputer,
    scaler,
    cat_imputer,
    encoder,
    X_train_num,
    X_train_cat
) = fit_preprocessor(X)


# ============================================================
# 13. TRANSFORM PROVIDED TEST DATA
# ============================================================

X_test_num, X_test_cat = transform_data(
    X_test,
    num_imputer,
    scaler,
    cat_imputer,
    encoder
)


# ============================================================
# 14. CATEGORICAL CARDINALITIES
# ============================================================

categorical_cardinalities = []

for i in range(
    len(categorical_cols)
):

    cardinality = (
        X_train_cat[:, i].max() + 1
    )

    categorical_cardinalities.append(
        int(cardinality)
    )


# ============================================================
# 15. BUILD FINAL MODEL
# ============================================================

final_model = build_model(
    num_features=len(numerical_cols),
    categorical_cardinalities=
        categorical_cardinalities,
    hidden_units=best_params["hidden_units"],
    dropout=best_params["dropout"],
    learning_rate=best_params["learning_rate"],
    optimizer_name=best_params["optimizer_name"]
)

final_model.summary()


# ============================================================
# 16. FINAL TRAINING
# ============================================================

if best_params["optimizer_name"] == "batch_gd":

    final_batch_size = len(
        X_train_num
    )

else:

    final_batch_size = 256


early_stopping = tf.keras.callbacks.EarlyStopping(
    monitor="loss",
    patience=5,
    restore_best_weights=True
)

history = final_model.fit(
    {
        "numerical": X_train_num,
        "categorical": X_train_cat
    },
    y,
    epochs=50,
    batch_size=final_batch_size,
    callbacks=[early_stopping],
    verbose=1
)


# ============================================================
# 17. TEST PREDICTIONS
# ============================================================

test_prob = final_model.predict(
    {
        "numerical": X_test_num,
        "categorical": X_test_cat
    },
    verbose=0
).ravel()


# ============================================================
# 18. SELECT THRESHOLD FROM CROSS-VALIDATION
# ============================================================
# Threshold is NOT selected using test labels.


cv_oof_prob = np.zeros(
    len(y),
    dtype=np.float32
)

for fold, (train_idx, val_idx) in enumerate(
    skf.split(X, y),
    start=1
):

    X_fold_train = X.iloc[train_idx]
    X_fold_val = X.iloc[val_idx]

    y_fold_train = y.iloc[train_idx]

    (
        fold_num_imputer,
        fold_scaler,
        fold_cat_imputer,
        fold_encoder,
        X_fold_train_num,
        X_fold_train_cat
    ) = fit_preprocessor(
        X_fold_train
    )

    X_fold_val_num, X_fold_val_cat = transform_data(
        X_fold_val,
        fold_num_imputer,
        fold_scaler,
        fold_cat_imputer,
        fold_encoder
    )

    fold_cardinalities = []

    for i in range(
        len(categorical_cols)
    ):

        fold_cardinalities.append(
            int(
                X_fold_train_cat[:, i].max() + 1
            )
        )

    fold_model = build_model(
        num_features=len(numerical_cols),
        categorical_cardinalities=
            fold_cardinalities,
        hidden_units=best_params["hidden_units"],
        dropout=best_params["dropout"],
        learning_rate=best_params["learning_rate"],
        optimizer_name=best_params["optimizer_name"]
    )

    if best_params["optimizer_name"] == "batch_gd":
        fold_batch_size = len(
            X_fold_train_num
        )
    else:
        fold_batch_size = 256

    fold_model.fit(
        {
            "numerical": X_fold_train_num,
            "categorical": X_fold_train_cat
        },
        y_fold_train,
        epochs=50,
        batch_size=fold_batch_size,
        verbose=0
    )

    cv_oof_prob[val_idx] = fold_model.predict(
        {
            "numerical": X_fold_val_num,
            "categorical": X_fold_val_cat
        },
        verbose=0
    ).ravel()

    del fold_model

    tf.keras.backend.clear_session()


# ============================================================
# 19. FIND BEST F1 THRESHOLD
# ============================================================

thresholds = np.arange(
    0.05,
    0.96,
    0.01
)

threshold_results = []

for threshold in thresholds:

    cv_pred = (
        cv_oof_prob >= threshold
    ).astype(int)

    f1 = f1_score(
        y,
        cv_pred,
        zero_division=0
    )

    precision = precision_score(
        y,
        cv_pred,
        zero_division=0
    )

    recall = recall_score(
        y,
        cv_pred,
        zero_division=0
    )

    threshold_results.append(
        {
            "threshold": threshold,
            "f1": f1,
            "precision": precision,
            "recall": recall
        }
    )


threshold_df = pd.DataFrame(
    threshold_results
)

best_threshold_row = threshold_df.loc[
    threshold_df["f1"].idxmax()
]

best_threshold = (
    best_threshold_row["threshold"]
)

print(
    "\nSelected threshold:",
    best_threshold
)


# ============================================================
# 20. FINAL TEST METRICS
# ============================================================

if HAS_TEST_LABELS:

    test_pred = (
        test_prob >= best_threshold
    ).astype(int)

    test_roc_auc = roc_auc_score(
        y_test,
        test_prob
    )

    test_pr_auc = average_precision_score(
        y_test,
        test_prob
    )

    test_log_loss = log_loss(
        y_test,
        test_prob
    )

    test_accuracy = accuracy_score(
        y_test,
        test_pred
    )

    test_precision = precision_score(
        y_test,
        test_pred,
        zero_division=0
    )

    test_recall = recall_score(
        y_test,
        test_pred,
        zero_division=0
    )

    test_f1 = f1_score(
        y_test,
        test_pred,
        zero_division=0
    )

    print("\n==============================")
    print("FINAL TEST RESULTS")
    print("==============================")

    print(
        f"ROC-AUC   : {test_roc_auc:.5f}"
    )

    print(
        f"PR-AUC    : {test_pr_auc:.5f}"
    )

    print(
        f"Log Loss  : {test_log_loss:.5f}"
    )

    print(
        f"Accuracy  : {test_accuracy:.5f}"
    )

    print(
        f"Precision : {test_precision:.5f}"
    )

    print(
        f"Recall    : {test_recall:.5f}"
    )

    print(
        f"F1        : {test_f1:.5f}"
    )


# ============================================================
# 21. SAVE TEST PREDICTIONS
# ============================================================

submission = pd.DataFrame({
    "prediction": test_prob
})

submission.to_csv(
    "ctr_predictions.csv",
    index=False
)

print(
    "\nSaved: ctr_predictions.csv"
)


# ============================================================
# 22. TRAINING LOSS CURVE
# ============================================================

plt.figure(figsize=(8, 5))

plt.plot(
    history.history["loss"],
    label="Training Loss"
)

plt.xlabel("Epoch")
plt.ylabel("Binary Cross-Entropy")
plt.title("Training Loss")

plt.legend()
plt.grid(True)
plt.show()


# ============================================================
# 23. TRAINING ROC-AUC CURVE
# ============================================================

if "roc_auc" in history.history:

    plt.figure(figsize=(8, 5))

    plt.plot(
        history.history["roc_auc"],
        label="Training ROC-AUC"
    )

    plt.xlabel("Epoch")
    plt.ylabel("ROC-AUC")
    plt.title("Training ROC-AUC")

    plt.legend()
    plt.grid(True)
    plt.show()


# ============================================================
# 24. THRESHOLD VS F1
# ============================================================

plt.figure(figsize=(8, 5))

plt.plot(
    threshold_df["threshold"],
    threshold_df["f1"],
    label="F1"
)

plt.axvline(
    best_threshold,
    linestyle="--",
    label=f"Best threshold = {best_threshold:.2f}"
)

plt.xlabel("Classification Threshold")
plt.ylabel("F1")
plt.title("Threshold Selection")

plt.legend()
plt.grid(True)
plt.show()


# ============================================================
# 25. RELIABILITY / CALIBRATION PLOT
# ============================================================

"""from sklearn.calibration import calibration_curve

if HAS_TEST_LABELS:

    prob_true, prob_pred = calibration_curve(
        y_test,
        test_prob,
        n_bins=10,
        strategy="uniform"
    )

    plt.figure(figsize=(7, 7))

    plt.plot(
        prob_pred,
        prob_true,
        marker="o",
        label="Model"
    )

    plt.plot(
        [0, 1],
        [0, 1],
        linestyle="--",
        label="Perfect calibration"
    )

    plt.xlabel("Predicted Probability")
    plt.ylabel("Observed Frequency")
    plt.title("Reliability Diagram")

    plt.legend()
    plt.grid(True)
    plt.show()
"""

# ============================================================
# 26. SAVE GRID SEARCH RESULTS
# ============================================================

results_df.to_csv(
    "grid_search_results.csv",
    index=False
)

print(
    "Saved: grid_search_results.csv"
)
```

```python
# ============================================================
# 0. IMPORTS
# ============================================================
import pandas as pd
import numpy as np
import gc
import tensorflow as tf
import matplotlib.pyplot as plt

from sklearn.model_selection import StratifiedKFold, ParameterGrid
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OrdinalEncoder

from sklearn.metrics import (
    roc_auc_score,
    average_precision_score,
    accuracy_score,
    log_loss,
    precision_score,
    recall_score,
    f1_score
)

from sklearn.calibration import calibration_curve


# ============================================================
# 1. REPRODUCIBILITY
# ============================================================

SEED = 42

np.random.seed(SEED)
tf.random.set_seed(SEED)


# ============================================================
# 2. LOAD DATA
# ============================================================

TRAIN_PATH = (
    "/kaggle/input/datasets/"
    "chinmaysharma147/dataset-task2/train.csv"
)

TEST_PATH = (
    "/kaggle/input/datasets/"
    "chinmaysharma147/dataset-task2/test.csv"
)

train = pd.read_csv(TRAIN_PATH)
test = pd.read_csv(TEST_PATH)

print("Train shape:", train.shape)
print("Test shape :", test.shape)


# ============================================================
# 3. DISPLAY COLUMNS
# ============================================================

print("\nTRAIN COLUMNS:")
print(train.columns.tolist())


# ============================================================
# 4. SEPARATE FEATURES AND TARGET
# ============================================================

TARGET = "label"

X = train.drop(columns=[TARGET])
y = train[TARGET].astype("int32")

if TARGET in test.columns:

    X_test = test.drop(columns=[TARGET])
    y_test = test[TARGET].astype("int32")

    HAS_TEST_LABELS = True

else:

    X_test = test.copy()
    y_test = None

    HAS_TEST_LABELS = False


# ============================================================
# 5. FEATURE LISTS
# ============================================================
#
# 
#
# The assignment says:
#   13 numerical features
#   26 categorical features
#
# ============================================================

numerical_cols = [
    'integer_feature_1',
    'integer_feature_2',
    'integer_feature_3',
    'integer_feature_4',
    'integer_feature_5',
    'integer_feature_6',
    'integer_feature_7',
    'integer_feature_8',
    'integer_feature_9',
    'integer_feature_10',
    'integer_feature_11',
    'integer_feature_12',
    'integer_feature_13'
]

categorical_cols = [
    'categorical_feature_1',
    'categorical_feature_2',
    'categorical_feature_3',
    'categorical_feature_4',
    'categorical_feature_5',
    'categorical_feature_6',
    'categorical_feature_7',
    'categorical_feature_8',
    'categorical_feature_9',
    'categorical_feature_10',
    'categorical_feature_11',
    'categorical_feature_12',
    'categorical_feature_13',
    'categorical_feature_14',
    'categorical_feature_15',
    'categorical_feature_16',
    'categorical_feature_17',
    'categorical_feature_18',
    'categorical_feature_19',
    'categorical_feature_20',
    'categorical_feature_21',
    'categorical_feature_22',
    'categorical_feature_23',
    'categorical_feature_24',
    'categorical_feature_25',
    'categorical_feature_26'
]


# ============================================================
# 6. FEATURE VALIDATION
# ============================================================

assert len(numerical_cols) == 13, (
    "There must be exactly 13 numerical columns."
)

assert len(categorical_cols) == 26, (
    "There must be exactly 26 categorical columns."
)

all_features = numerical_cols + categorical_cols

assert len(all_features) == len(set(all_features)), (
    "Duplicate column names detected."
)

missing_from_train = [
    col for col in all_features
    if col not in X.columns
]

missing_from_test = [
    col for col in all_features
    if col not in X_test.columns
]

if missing_from_train:

    raise ValueError(
        "These feature columns are missing from TRAIN:\n"
        + str(missing_from_train)
    )

if missing_from_test:

    raise ValueError(
        "These feature columns are missing from TEST:\n"
        + str(missing_from_test)
    )

print("\nFeature validation successful.")
print("Numerical features :", len(numerical_cols))
print("Categorical features:", len(categorical_cols))


# ============================================================
# 7. HELPER FUNCTION FOR MODEL INPUTS
# ============================================================
#
# TensorFlow model has inputs:
#
# numerical_input
# cat_0
# cat_1
# ...
# cat_25
#
# Therefore we must construct the dictionary using EXACTLY
# these names.
#
# ============================================================

def make_model_inputs(X_num, X_cat):

    inputs = {
        "numerical_input": X_num
    }

    for i in range(X_cat.shape[1]):

        inputs[f"cat_{i}"] = X_cat[:, i]

    return inputs


# ============================================================
# 8. PREPROCESSING
# ============================================================

def fit_preprocessor(X_train):
    """
    Fit preprocessing ONLY on the supplied training data.

    Numerical:
        NaN -> median
        standardization

    Categorical:
        NaN -> MISSING
        ordinal encoding
        unknown category -> 0
    """

    # ========================================================
    # NUMERICAL
    # ========================================================

    num_imputer = SimpleImputer(
        strategy="median"
    )

    X_train_num = num_imputer.fit_transform(
        X_train[numerical_cols]
    )

    scaler = StandardScaler()

    X_train_num = scaler.fit_transform(
        X_train_num
    )


    # ========================================================
    # CATEGORICAL
    # ========================================================

    # IMPORTANT:
    # Fill NaN BEFORE converting to string.
    X_train_cat_raw = (
        X_train[categorical_cols]
        .fillna("MISSING")
        .astype(str)
    )

    cat_imputer = SimpleImputer(
        strategy="constant",
        fill_value="MISSING"
    )

    X_train_cat = cat_imputer.fit_transform(
        X_train_cat_raw
    )

    encoder = OrdinalEncoder(
        handle_unknown="use_encoded_value",
        unknown_value=-1
    )

    X_train_cat = encoder.fit_transform(
        X_train_cat
    )

    # --------------------------------------------------------
    # Shift categories by +1
    #
    # Original:
    # unknown = -1
    # known   = 0,1,2,...
    #
    # After shift:
    # unknown = 0
    # known   = 1,2,3,...
    #
    # This is useful for embeddings because 0 is now a valid
    # explicit unknown token.
    # --------------------------------------------------------

    X_train_cat = (
        X_train_cat.astype("int32") + 1
    )


    return (
        num_imputer,
        scaler,
        cat_imputer,
        encoder,
        X_train_num.astype("float32"),
        X_train_cat.astype("int32")
    )


# ============================================================
# 9. TRANSFORM DATA
# ============================================================

def transform_data(
    X_data,
    num_imputer,
    scaler,
    cat_imputer,
    encoder
):
    """
    Transform data using preprocessing already fitted
    on training data.
    """

    # ========================================================
    # NUMERICAL
    # ========================================================

    X_num = num_imputer.transform(
        X_data[numerical_cols]
    )

    X_num = scaler.transform(
        X_num
    )


    # ========================================================
    # CATEGORICAL
    # ========================================================

    X_cat_raw = (
        X_data[categorical_cols]
        .fillna("MISSING")
        .astype(str)
    )

    X_cat = cat_imputer.transform(
        X_cat_raw
    )

    X_cat = encoder.transform(
        X_cat
    )

    X_cat = (
        X_cat.astype("int32") + 1
    )


    return (
        X_num.astype("float32"),
        X_cat.astype("int32")
    )


# ============================================================
# 10. CROSS NETWORK
# ============================================================

class CrossNetwork(tf.keras.layers.Layer):
    """
    Deep & Cross Network explicit feature crossing.

    Formula:

        x_(l+1)
        =
        x_0 * (x_l^T w_l)
        +
        b_l
        +
        x_l

    where:

        x_0 = original feature representation
        x_l = current cross representation
        w_l = trainable weight vector
        b_l = trainable bias
    """

    def __init__(
        self,
        num_layers=2,
        **kwargs
    ):

        super().__init__(**kwargs)

        self.num_layers = num_layers


    def build(self, input_shape):

        feature_dim = int(
            input_shape[-1]
        )

        self.weights_cross = []
        self.biases_cross = []


        for i in range(self.num_layers):

            w = self.add_weight(
                name=f"cross_weight_{i}",
                shape=(feature_dim, 1),
                initializer="glorot_uniform",
                trainable=True
            )

            b = self.add_weight(
                name=f"cross_bias_{i}",
                shape=(feature_dim,),
                initializer="zeros",
                trainable=True
            )

            self.weights_cross.append(w)
            self.biases_cross.append(b)
            
        super().build(input_shape)

    def call(self, x0):

        x = x0


        for w, b in zip(
            self.weights_cross,
            self.biases_cross
        ):

            # x_l^T w_l
            scalar = tf.matmul(
                x,
                w
            )

            # x_0 * (x_l^T w_l)
            cross_term = (
                x0 * scalar
            )

            # x_(l+1)
            x = (
                cross_term
                + b
                + x
            )


        return x


# ============================================================
# 11. BUILD DEEP + CROSS MODEL
# ============================================================

def build_model(
    num_numerical_features,
    categorical_cardinalities,
    hidden_units=(128, 64),
    dropout=0.3,
    learning_rate=1e-3,
    optimizer_name="adam",
    cross_layers=2
):

    # ========================================================
    # NUMERICAL INPUT
    # ========================================================

    numerical_input = tf.keras.Input(
        shape=(num_numerical_features,),
        name="numerical_input"
    )


    # ========================================================
    # CATEGORICAL INPUTS + EMBEDDINGS
    # ========================================================

    categorical_inputs = []
    embeddings = []


    for i, cardinality in enumerate(
        categorical_cardinalities
    ):

        cat_input = tf.keras.Input(
            shape=(1,),
            dtype=tf.int32,
            name=f"cat_{i}"
        )


        # ----------------------------------------------------
        # Embedding dimension
        #
        # sqrt(cardinality) is used as a small heuristic.
        # Minimum = 4
        # Maximum = 32
        # ----------------------------------------------------

        embedding_dim = min(
            32,
            max(
                4,
                int(np.sqrt(cardinality))
            )
        )


        embedding = tf.keras.layers.Embedding(
            input_dim=cardinality,
            output_dim=embedding_dim,
            name=f"embedding_{i}"
        )(cat_input)


        embedding = tf.keras.layers.Flatten()(
            embedding
        )


        categorical_inputs.append(
            cat_input
        )

        embeddings.append(
            embedding
        )


    # ========================================================
    # COMMON FEATURE REPRESENTATION
    # ========================================================

    x = tf.keras.layers.Concatenate(
        name="feature_representation"
    )(
        [numerical_input] + embeddings
    )


    # ========================================================
    # DEEP BRANCH
    # ========================================================

    deep = x


    for units in hidden_units:

        deep = tf.keras.layers.Dense(
            units,
            activation="relu"
        )(deep)


        deep = tf.keras.layers.BatchNormalization()(
            deep
        )


        deep = tf.keras.layers.Dropout(
            dropout
        )(deep)


    # ========================================================
    # CROSS BRANCH
    # ========================================================
    #
    # NO use_cross SWITCH.
    #
    # The architecture is ALWAYS:
    #
    #       Deep + Cross
    #
    # ========================================================

    cross = CrossNetwork(
        num_layers=cross_layers,
        name="cross_network"
    )(x)


    # ========================================================
    # COMBINE DEEP + CROSS
    # ========================================================

    combined = tf.keras.layers.Concatenate(
        name="deep_cross_concat"
    )(
        [deep, cross]
    )


    # ========================================================
    # OUTPUT
    # ========================================================

    logit = tf.keras.layers.Dense(
        1,
        activation=None,
        name="logit"
    )(combined)


    output = tf.keras.layers.Activation(
        "sigmoid",
        name="probability"
    )(logit)


    # ========================================================
    # COMPLETE MODEL
    # ========================================================

    model = tf.keras.Model(
        inputs=[
            numerical_input
        ] + categorical_inputs,
        outputs=output
    )


    # ========================================================
    # OPTIMIZER
    # ========================================================

    if optimizer_name == "adam":

        optimizer = tf.keras.optimizers.Adam(
            learning_rate=learning_rate
        )


    elif optimizer_name == "batch_gd":

        optimizer = tf.keras.optimizers.SGD(
            learning_rate=learning_rate
        )


    else:

        raise ValueError(
            f"Unknown optimizer: {optimizer_name}"
        )


    # ========================================================
    # COMPILE
    # ========================================================

    model.compile(
        optimizer=optimizer,
        loss="binary_crossentropy",
        metrics=[
            tf.keras.metrics.AUC(
                name="roc_auc"
            )
        ]
    )


    return model


# ============================================================
# 12. HYPERPARAMETER GRID
# ============================================================
#
# This searches:
#
# hidden architecture
# dropout
# learning rate
# optimizer
# batch size
#
# Cross network is always present.
#
# cross_layers = 2 is kept fixed here to control computation.
#
# If you want to investigate cross depth later, use:
#
#     "cross_layers": [1, 2, 3]
#
# ============================================================

param_grid = {

    "hidden_units": [
        (64,),
        (128, 64),
        (256, 128, 64)
    ],

    "dropout": [
        0.0,
        0.3
    ],

    "learning_rate": [
        1e-3,
        1e-4
    ],

    "optimizer_name": [
        "adam",
        "batch_gd"
    ],

    "batch_size": [
        256
    ],

    "cross_layers": [
        1,
        2,
        3
    ]
}


# ============================================================
# 13. STRATIFIED K-FOLD
# ============================================================

N_SPLITS = 5

skf = StratifiedKFold(
    n_splits=N_SPLITS,
    shuffle=True,
    random_state=SEED
)


# ============================================================
# 14. GRID SEARCH
# ============================================================

grid_results = []

param_combinations = list(
    ParameterGrid(param_grid)
)

print(
    "\nTotal hyperparameter combinations:",
    len(param_combinations)
)

print(
    "Total model fits:",
    len(param_combinations) * N_SPLITS
)


# ============================================================
# 15. GRID SEARCH LOOP
# ============================================================

for combination_number, params in enumerate(
    param_combinations,
    start=1
):

    print("\n" + "=" * 80)

    print(
        f"Combination "
        f"{combination_number}/"
        f"{len(param_combinations)}"
    )

    print(params)

    print("=" * 80)


    fold_scores = []


    # ========================================================
    # K-FOLD
    # ========================================================

    for fold, (train_idx, val_idx) in enumerate(
        skf.split(X, y),
        start=1
    ):

        print(
            f"\nFold {fold}/{N_SPLITS}"
        )


        # ----------------------------------------------------
        # Split
        # ----------------------------------------------------

        X_fold_train = X.iloc[
            train_idx
        ]

        X_fold_val = X.iloc[
            val_idx
        ]

        y_fold_train = y.iloc[
            train_idx
        ]

        y_fold_val = y.iloc[
            val_idx
        ]


        # ----------------------------------------------------
        # Fold-specific preprocessing
        # ----------------------------------------------------

        (
            num_imputer,
            scaler,
            cat_imputer,
            encoder,
            X_fold_train_num,
            X_fold_train_cat
        ) = fit_preprocessor(
            X_fold_train
        )


        X_fold_val_num, X_fold_val_cat = (
            transform_data(
                X_fold_val,
                num_imputer,
                scaler,
                cat_imputer,
                encoder
            )
        )


        # ----------------------------------------------------
        # Category cardinalities
        # ----------------------------------------------------

        categorical_cardinalities = []


        for i in range(
            len(categorical_cols)
        ):

            cardinality = int(
                X_fold_train_cat[:, i].max()
                + 1
            )

            categorical_cardinalities.append(
                cardinality
            )


        # ----------------------------------------------------
        # Build model
        # ----------------------------------------------------

        model = build_model(

            num_numerical_features=
                len(numerical_cols),

            categorical_cardinalities=
                categorical_cardinalities,

            hidden_units=
                params["hidden_units"],

            dropout=
                params["dropout"],

            learning_rate=
                params["learning_rate"],

            optimizer_name=
                params["optimizer_name"],

            cross_layers=
                params["cross_layers"]
        )


        # ----------------------------------------------------
        # Batch size
        # ----------------------------------------------------

        if params["optimizer_name"] == "batch_gd":

            batch_size = len(
                X_fold_train_num
            )

        else:

            batch_size = params[
                "batch_size"
            ]


        # ----------------------------------------------------
        # Early stopping
        # ----------------------------------------------------

        early_stopping = (
            tf.keras.callbacks.EarlyStopping(

                monitor="val_loss",

                patience=5,

                restore_best_weights=True
            )
        )


        # ----------------------------------------------------
        # Train
        # ----------------------------------------------------

        history = model.fit(

            make_model_inputs(
                X_fold_train_num,
                X_fold_train_cat
            ),

            y_fold_train,

            validation_data=(

                make_model_inputs(
                    X_fold_val_num,
                    X_fold_val_cat
                ),

                y_fold_val
            ),

            epochs=50,

            batch_size=batch_size,

            callbacks=[
                early_stopping
            ],

            verbose=0
        )


        # ----------------------------------------------------
        # Validation prediction
        # ----------------------------------------------------

        val_prob = model.predict(

            make_model_inputs(
                X_fold_val_num,
                X_fold_val_cat
            ),

            verbose=0
        ).ravel()


        # ----------------------------------------------------
        # ROC-AUC
        # ----------------------------------------------------

        fold_auc = roc_auc_score(
            y_fold_val,
            val_prob
        )

        fold_scores.append(
            fold_auc
        )


        print(
            f"Fold {fold} ROC-AUC: "
            f"{fold_auc:.5f}"
        )


        # ----------------------------------------------------
        # Cleanup
        # ----------------------------------------------------

        del model
        del history

        tf.keras.backend.clear_session()

        gc.collect()


    # ========================================================
    # MEAN CV SCORE
    # ========================================================

    mean_auc = np.mean(
        fold_scores
    )

    std_auc = np.std(
        fold_scores
    )


    result = params.copy()

    result["mean_roc_auc"] = (
        mean_auc
    )

    result["std_roc_auc"] = (
        std_auc
    )


    grid_results.append(
        result
    )


    print(
        f"\nMean ROC-AUC: "
        f"{mean_auc:.5f} "
        f"± {std_auc:.5f}"
    )


# ============================================================
# 16. GRID SEARCH RESULTS
# ============================================================

results_df = pd.DataFrame(
    grid_results
).sort_values(
    "mean_roc_auc",
    ascending=False
).reset_index(
    drop=True
)


print(
    "\n\n=============================="
)

print(
    "GRID SEARCH RESULTS"
)

print(
    "=============================="
)

print(
    results_df.to_string(
        index=False
    )
)


# ============================================================
# 17. BEST PARAMETERS
# ============================================================

best_params = (
    results_df.iloc[0].to_dict()
)


print(
    "\n\n=============================="
)

print(
    "BEST PARAMETERS"
)

print(
    "=============================="
)

print(
    "Hidden units:",
    best_params["hidden_units"]
)

print(
    "Dropout:",
    best_params["dropout"]
)

print(
    "Learning rate:",
    best_params["learning_rate"]
)

print(
    "Optimizer:",
    best_params["optimizer_name"]
)

print(
    "Cross layers:",
    best_params["cross_layers"]
)

print(
    "CV ROC-AUC:",
    best_params["mean_roc_auc"]
)


# ============================================================
# 18. FIT PREPROCESSING ON ALL TRAINING DATA
# ============================================================
#
# This is done ONLY after hyperparameter selection.
#
# The test set is still untouched.
#
# ============================================================

(
    num_imputer,
    scaler,
    cat_imputer,
    encoder,
    X_train_num,
    X_train_cat
) = fit_preprocessor(
    X
)


# ============================================================
# 19. TRANSFORM PROVIDED TEST DATA
# ============================================================

X_test_num, X_test_cat = transform_data(

    X_test,

    num_imputer,
    scaler,
    cat_imputer,
    encoder
)


# ============================================================
# 20. CATEGORICAL CARDINALITIES
# ============================================================

categorical_cardinalities = []


for i in range(
    len(categorical_cols)
):

    cardinality = int(
        X_train_cat[:, i].max()
        + 1
    )

    categorical_cardinalities.append(
        cardinality
    )


print(
    "\nCategorical cardinalities:"
)

print(
    categorical_cardinalities
)


# ============================================================
# 21. FINAL MODEL
# ============================================================

final_model = build_model(

    num_numerical_features=
        len(numerical_cols),

    categorical_cardinalities=
        categorical_cardinalities,

    hidden_units=
        best_params["hidden_units"],

    dropout=
        best_params["dropout"],

    learning_rate=
        best_params["learning_rate"],

    optimizer_name=
        best_params["optimizer_name"],

    cross_layers=
        int(best_params["cross_layers"])
)


final_model.summary()


# ============================================================
# 22. FINAL TRAINING
# ============================================================
#
# Since the final model uses ALL provided training data,
# there is no independent validation set here.
#
# Therefore we do not monitor val_loss.
#
# The CV stage already performed early stopping and model
# selection.
#
# ============================================================

if best_params["optimizer_name"] == "batch_gd":

    final_batch_size = len(
        X_train_num
    )

else:

    final_batch_size = int(
        best_params["batch_size"]
    )


history = final_model.fit(

    make_model_inputs(
        X_train_num,
        X_train_cat
    ),

    y,

    epochs=50,

    batch_size=final_batch_size,

    verbose=1
)


# ============================================================
# 23. FINAL TRAINING CURVE
# ============================================================

plt.figure(figsize=(8, 5))

plt.plot(
    history.history["loss"],
    label="Training Loss"
)

plt.xlabel("Epoch")
plt.ylabel("Binary Cross-Entropy Loss")

plt.title(
    "Final Model Training Loss"
)

plt.legend()
plt.grid()

plt.show()


# ============================================================
# 24. TEST PREDICTIONS
# ============================================================

test_prob = final_model.predict(

    make_model_inputs(
        X_test_num,
        X_test_cat
    ),

    verbose=0
).ravel()


print(
    "\nTest predictions generated."
)

print(
    "Prediction range:",
    test_prob.min(),
    "to",
    test_prob.max()
)


# ============================================================
# 25. OOF PREDICTIONS
# ============================================================
#
# We need out-of-fold predictions to select the classification
# threshold WITHOUT using test labels.
#
# ============================================================

cv_oof_prob = np.zeros(
    len(y),
    dtype=np.float32
)


# Store OOF training histories for visualization
oof_histories = []


# ============================================================
# 26. OOF LOOP
# ============================================================

for fold, (train_idx, val_idx) in enumerate(

    skf.split(X, y),

    start=1
):

    print(
        f"\nOOF Fold "
        f"{fold}/{N_SPLITS}"
    )


    # --------------------------------------------------------
    # Split
    # --------------------------------------------------------

    X_fold_train = X.iloc[
        train_idx
    ]

    X_fold_val = X.iloc[
        val_idx
    ]

    y_fold_train = y.iloc[
        train_idx
    ]

    y_fold_val = y.iloc[
        val_idx
    ]


    # --------------------------------------------------------
    # Fold preprocessing
    # --------------------------------------------------------

    (
        fold_num_imputer,
        fold_scaler,
        fold_cat_imputer,
        fold_encoder,
        X_fold_train_num,
        X_fold_train_cat
    ) = fit_preprocessor(
        X_fold_train
    )


    X_fold_val_num, X_fold_val_cat = (
        transform_data(

            X_fold_val,

            fold_num_imputer,

            fold_scaler,

            fold_cat_imputer,

            fold_encoder
        )
    )


    # --------------------------------------------------------
    # Fold cardinalities
    # --------------------------------------------------------

    fold_cardinalities = []


    for i in range(
        len(categorical_cols)
    ):

        cardinality = int(
            X_fold_train_cat[:, i].max()
            + 1
        )

        fold_cardinalities.append(
            cardinality
        )


    # --------------------------------------------------------
    # Build fold model
    # --------------------------------------------------------

    fold_model = build_model(

        num_numerical_features=
            len(numerical_cols),

        categorical_cardinalities=
            fold_cardinalities,

        hidden_units=
            best_params["hidden_units"],

        dropout=
            best_params["dropout"],

        learning_rate=
            best_params["learning_rate"],

        optimizer_name=
            best_params["optimizer_name"],

        cross_layers=
            int(best_params["cross_layers"])
    )


    # --------------------------------------------------------
    # Batch size
    # --------------------------------------------------------

    if best_params["optimizer_name"] == "batch_gd":

        fold_batch_size = len(
            X_fold_train_num
        )

    else:

        fold_batch_size = int(
            best_params["batch_size"]
        )


    # --------------------------------------------------------
    # Early stopping
    # --------------------------------------------------------

    fold_early_stopping = (
        tf.keras.callbacks.EarlyStopping(

            monitor="val_loss",

            patience=5,

            restore_best_weights=True
        )
    )


    # --------------------------------------------------------
    # Train
    # --------------------------------------------------------

    fold_history = fold_model.fit(

        make_model_inputs(
            X_fold_train_num,
            X_fold_train_cat
        ),

        y_fold_train,

        validation_data=(

            make_model_inputs(
                X_fold_val_num,
                X_fold_val_cat
            ),

            y_fold_val
        ),

        epochs=50,

        batch_size=fold_batch_size,

        callbacks=[
            fold_early_stopping
        ],

        verbose=0
    )


    oof_histories.append(
        fold_history.history
    )


    # --------------------------------------------------------
    # OOF prediction
    # --------------------------------------------------------

    val_prob = fold_model.predict(

        make_model_inputs(
            X_fold_val_num,
            X_fold_val_cat
        ),

        verbose=0
    ).ravel()


    cv_oof_prob[
        val_idx
    ] = val_prob


    fold_auc = roc_auc_score(
        y_fold_val,
        val_prob
    )


    print(
        f"Fold {fold} OOF ROC-AUC: "
        f"{fold_auc:.5f}"
    )


    # --------------------------------------------------------
    # Cleanup
    # --------------------------------------------------------

    del fold_model
    del fold_history

    tf.keras.backend.clear_session()

    gc.collect()


# ============================================================
# 27. OOF ROC-AUC
# ============================================================

oof_roc_auc = roc_auc_score(
    y,
    cv_oof_prob
)

oof_pr_auc = average_precision_score(
    y,
    cv_oof_prob
)

oof_log_loss = log_loss(
    y,
    cv_oof_prob
)


print(
    "\n=============================="
)

print(
    "OOF PERFORMANCE"
)

print(
    "=============================="
)

print(
    f"OOF ROC-AUC : {oof_roc_auc:.5f}"
)

print(
    f"OOF PR-AUC  : {oof_pr_auc:.5f}"
)

print(
    f"OOF Log Loss: {oof_log_loss:.5f}"
)


# ============================================================
# 28. FIND BEST CLASSIFICATION THRESHOLD
# ============================================================
#
# IMPORTANT:
#
# Threshold is selected using OOF predictions.
#
# Test labels are NOT used.
#
# ============================================================

thresholds = np.arange(
    0.05,
    0.96,
    0.01
)


threshold_results = []


for threshold in thresholds:

    oof_pred = (
        cv_oof_prob >= threshold
    ).astype(int)


    f1 = f1_score(
        y,
        oof_pred,
        zero_division=0
    )


    precision = precision_score(
        y,
        oof_pred,
        zero_division=0
    )


    recall = recall_score(
        y,
        oof_pred,
        zero_division=0
    )


    accuracy = accuracy_score(
        y,
        oof_pred
    )


    threshold_results.append({

        "threshold": threshold,

        "f1": f1,

        "precision": precision,

        "recall": recall,

        "accuracy": accuracy
    })


threshold_df = pd.DataFrame(
    threshold_results
)


best_threshold_row = (
    threshold_df.loc[
        threshold_df["f1"].idxmax()
    ]
)


best_threshold = float(
    best_threshold_row["threshold"]
)


print(
    "\n=============================="
)

print(
    "THRESHOLD SELECTION"
)

print(
    "=============================="
)

print(
    f"Best threshold: "
    f"{best_threshold:.2f}"
)

print(
    f"OOF F1       : "
    f"{best_threshold_row['f1']:.5f}"
)

print(
    f"OOF Precision: "
    f"{best_threshold_row['precision']:.5f}"
)

print(
    f"OOF Recall   : "
    f"{best_threshold_row['recall']:.5f}"
)


# ============================================================
# 29. THRESHOLD CURVE
# ============================================================

plt.figure(figsize=(8, 5))

plt.plot(
    threshold_df["threshold"],
    threshold_df["f1"],
    label="F1"
)

plt.plot(
    threshold_df["threshold"],
    threshold_df["precision"],
    label="Precision"
)

plt.plot(
    threshold_df["threshold"],
    threshold_df["recall"],
    label="Recall"
)

plt.axvline(
    best_threshold,
    linestyle="--",
    label=f"Best threshold = {best_threshold:.2f}"
)

plt.xlabel("Classification Threshold")
plt.ylabel("Metric")

plt.title(
    "Threshold Selection Using OOF Predictions"
)

plt.legend()
plt.grid()

plt.show()


# ============================================================
# 30. OOF TRAINING / VALIDATION CURVES
# ============================================================
#
# Display the first fold's training history.
#
# The fold histories are also available in:
#
#     oof_histories
#
# ============================================================

if len(oof_histories) > 0:

    first_history = oof_histories[0]


    plt.figure(figsize=(8, 5))

    plt.plot(
        first_history["loss"],
        label="Training Loss"
    )

    plt.plot(
        first_history["val_loss"],
        label="Validation Loss"
    )

    plt.xlabel("Epoch")
    plt.ylabel("Binary Cross-Entropy Loss")

    plt.title(
        "Training vs Validation Loss - OOF Fold 1"
    )

    plt.legend()
    plt.grid()

    plt.show()


    # --------------------------------------------------------
    # ROC-AUC curve
    # --------------------------------------------------------

    if (
        "roc_auc" in first_history
        and
        "val_roc_auc" in first_history
    ):

        plt.figure(figsize=(8, 5))

        plt.plot(
            first_history["roc_auc"],
            label="Training ROC-AUC"
        )

        plt.plot(
            first_history["val_roc_auc"],
            label="Validation ROC-AUC"
        )

        plt.xlabel("Epoch")
        plt.ylabel("ROC-AUC")

        plt.title(
            "Training vs Validation ROC-AUC - OOF Fold 1"
        )

        plt.legend()
        plt.grid()

        plt.show()


# ============================================================
# 31. CALIBRATION / RELIABILITY CURVE
# ============================================================
#
# Use OOF predictions rather than test predictions.
#
# Perfect calibration:
#
#       predicted probability = actual frequency
#
# ============================================================

prob_true, prob_pred = calibration_curve(

    y,

    cv_oof_prob,

    n_bins=10,

    strategy="uniform"
)


plt.figure(figsize=(7, 7))

plt.plot(
    prob_pred,
    prob_true,
    marker="o",
    label="DCN"
)

plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--",
    label="Perfect calibration"
)

plt.xlabel(
    "Mean predicted probability"
)

plt.ylabel(
    "Fraction of positives"
)

plt.title(
    "Reliability Diagram - OOF Predictions"
)

plt.legend()
plt.grid()

plt.show()


# ============================================================
# 32. FINAL TEST METRICS
# ============================================================

if HAS_TEST_LABELS:

    # --------------------------------------------------------
    # Apply OOF-selected threshold
    # --------------------------------------------------------

    test_pred = (
        test_prob >= best_threshold
    ).astype(int)


    # --------------------------------------------------------
    # Probability metrics
    # --------------------------------------------------------

    test_roc_auc = roc_auc_score(
        y_test,
        test_prob
    )


    test_pr_auc = average_precision_score(
        y_test,
        test_prob
    )


    test_log_loss = log_loss(
        y_test,
        test_prob
    )


    # --------------------------------------------------------
    # Threshold metrics
    # --------------------------------------------------------

    test_accuracy = accuracy_score(
        y_test,
        test_pred
    )


    test_precision = precision_score(
        y_test,
        test_pred,
        zero_division=0
    )


    test_recall = recall_score(
        y_test,
        test_pred,
        zero_division=0
    )


    test_f1 = f1_score(
        y_test,
        test_pred,
        zero_division=0
    )


    # ========================================================
    # PRINT FINAL RESULTS
    # ========================================================

    print(
        "\n\n"
        "========================================"
    )

    print(
        "FINAL TEST RESULTS"
    )

    print(
        "========================================"
    )


    print(
        f"ROC-AUC   : "
        f"{test_roc_auc:.5f}"
    )


    print(
        f"PR-AUC    : "
        f"{test_pr_auc:.5f}"
    )


    print(
        f"Log Loss  : "
        f"{test_log_loss:.5f}"
    )


    print(
        f"Accuracy  : "
        f"{test_accuracy:.5f}"
    )


    print(
        f"Precision : "
        f"{test_precision:.5f}"
    )


    print(
        f"Recall    : "
        f"{test_recall:.5f}"
    )


    print(
        f"F1        : "
        f"{test_f1:.5f}"
    )


    print(
        f"Threshold : "
        f"{best_threshold:.2f}"
    )


else:

    print(
        "\nTest labels are not available."
    )

    print(
        "Test probabilities have been generated,"
        " but supervised test metrics cannot be calculated."
    )


# ============================================================
# 33. OPTIONAL: SAVE TEST PREDICTIONS
# ============================================================

submission = pd.DataFrame({

    "prediction_probability": test_prob,

    "prediction": (
        test_prob >= best_threshold
    ).astype(int)

})


submission.to_csv(
    "/kaggle/working/dcn_predictions.csv",
    index=False
)


print(
    "\nPrediction file saved to:"
)

print(
    "/kaggle/working/dcn_predictions.csv"
)
```
